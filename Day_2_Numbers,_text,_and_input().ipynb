**Exercise 1 : Ask the user for their name, then print Hello, \<name>! Welcome to Python.**

```python
user_name = str(input("Please Enter Your Name:"))
print(f"Hello, {user_name}! Welcome to Python.")
```

**Exercise 2 : Ask for two whole numbers. Print their sum, difference, product, and division in a clear sentence each, such as The sum is 17.**

```python
user_number1 = int(input("Please Enter The First Number:"))
user_number2 = int(input("Please Enter The Second Number"))
sum = user_number1 + user_number2
difference = user_number1 - user_number2
product = user_number1 * user_number2
division = user_number1 / user_number2
print(f"The Sum of {user_number1} + {user_number2} = {sum} ")
print(f"The Difference of {user_number1} - {user_number2} = {difference} ")
print(f"The Product of {user_number1} * {user_number2} = {product} ")
print(f"The Division of {user_number1} / {user_number2} = {division} ")
```

**Exercise 3 : Ask for the user’s city. Print it in uppercase, then print how many characters it has, then print its first letter.**

```python
user_city = str(input("Please Enter Your City:"))
print(f"{user_city} in uppercase :" ,user_city.upper())
print(f"{user_city} has" ,len(user_city),"characters.")
print(f"The first letter in {user_city} is",user_city[0])
```

**Exercise 4 : A coffee shop calculator. Ask for the price of one coffee (a decimal number) and how many the user wants. Print Total cost: X.XX USD with exactly 2 decimals.**

```python
coffee_price = float(input("Enter the Coffee price:"))
user_amount = int(input("How many cups of Coffee do you want?"))
total = coffee_price * user_amount
print(f"Total cost :{total:.2f}")
```

**Exercise 5 : Ask for a number of minutes. Print how many hours and how many leftover minutes it equals. For example, 135 minutes → 2 hours and 15 minutes.**

```python
user_minutes = int(input("Please enter a number of minutes:"))
print(f"{user_minutes} covert to hours is equal to",user_minutes // 60,"hours and",user_minutes % 60,"minutes.")
```
